```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
file_path = r"C:\Files\Excel\Project\Stroke_Prediction_Dataset.xlsx" 
df = pd.read_excel(file_path) 
df.head() 
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize=(8,5)) 
plt.hist(df['Age'], bins=20) 
plt.title("Age Distribution of Patients") 
plt.xlabel("Age") 
plt.ylabel("Number of Patients") 
plt.show() 
```

```python
stroke_patients = df[df['Stroke History'] == 1]
```

```python
plt.figure(figsize=(8,5)) 
stroke_patients['Work Type'].value_counts().plot(kind='bar') 
plt.title("Work Type Distribution Among Stroke Patients") 
plt.xlabel("Work Type") 
plt.ylabel("Number of Patients") 
plt.show() 
```

```python
plt.figure(figsize=(8,5)) 
plt.scatter(stroke_patients['Age'], stroke_patients['Body Mass Index (BMI)']) 
plt.title("Age vs BMI for Stroke Patients") 
plt.xlabel("Age") 
plt.ylabel("BMI") 
plt.show() 
```

```python
corr_data = stroke_patients[['Age', 'Body Mass Index (BMI)', 'Average Glucose Level']] 
correlation_matrix = corr_data.corr() 
```

```python
plt.figure(figsize=(6,5)) 
sns.heatmap(correlation_matrix, annot=True) 
plt.title("Correlation Heatmap of Stroke Patient Variables") 
plt.show() 
```
