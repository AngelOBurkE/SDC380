```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
file_path = r"C:\Files\Excel\Project\Happiness_Development_Index.xlsx" 
df = pd.read_excel(file_path) 
df.head() 
```

```python
df.isnull().sum()
```

```python
plt.figure() 
plt.hist(df['Human Development Index (HDI)'], bins=20) 
plt.xlabel('HDI Score') 
plt.ylabel('Number of Countries') 
plt.title('Distribution of Human Development Index (HDI)') 
plt.show() 
```

```python
top10_hdi = df.sort_values(by='Human Development Index (HDI)', ascending=False).head(10) 
plt.figure() 
plt.barh(top10_hdi['Country'], top10_hdi['Human Development Index (HDI)']) 
plt.xlabel('HDI Score') 
plt.ylabel('Country') 
plt.title('Top 10 Countries by HDI') 
plt.gca().invert_yaxis() 
plt.show() 
```

```python
plt.figure() 
plt.scatter(df['Gross national income (GNI) per capita'], df['Human Development Index (HDI)']) 
plt.xlabel('GNI per Capita') 
plt.ylabel('HDI') 
plt.title('HDI vs Gross National Income per Capita') 
plt.show() 
```

```python
numerical_cols = [ 
    'Human Development Index (HDI)', 
    'Life expectancy at birth', 
    'Expected years of schooling', 
    'Mean years of schooling', 
    'Gross national income (GNI) per capita' 
] 
corr_matrix = df[numerical_cols].corr() 
plt.figure() 
sns.heatmap(corr_matrix, annot=True) 
plt.title('Correlation Heatmap of HDI and Related Variables') 
plt.show() 
```
