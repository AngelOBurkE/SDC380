```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
file_path = r"C:\Files\Excel\Project\StudentsPerformance.xlsx" 
df = pd.read_excel(file_path) 
df.head() 
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize=(8,5)) 
plt.hist(df['math score'], bins=20, color='skyblue', edgecolor='black') 
plt.title("Distribution of Math Scores") 
plt.xlabel("Math Score") 
plt.ylabel("Number of Students") 
plt.show() 
```

```python
avg_math_by_race = df.groupby('race/ethnicity')['math score'].mean().sort_values() 
plt.figure(figsize=(8,5)) 
avg_math_by_race.plot(kind='bar', color='lightgreen') 
plt.title("Average Math Scores by Race/Ethnicity") 
plt.xlabel("Race/Ethnicity") 
plt.ylabel("Average Math Score") 
plt.show() 
```

```python
plt.figure(figsize=(8,5)) 
plt.scatter(df['math score'], df['writing score'], alpha=0.6, color='salmon') 
plt.title("Math Score vs Writing Score") 
plt.xlabel("Math Score") 
plt.ylabel("Writing Score") 
plt.show() 
```

```python
numerical_cols = ['math score', 'reading score', 'writing score'] 
corr_matrix = df[numerical_cols].corr() 
plt.figure(figsize=(6,5)) 
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm') 
plt.title("Correlation Heatmap of Student Scores") 
plt.show() 
```
